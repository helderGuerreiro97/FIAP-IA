```r
# ============================================================
# 1. BIBLIOTECAS E CONFIGURAÇÃO DA API
# ============================================================
library(httr)
library(jsonlite)

api_key <- "4bfeeb5809135a2a28b59273ef2b3767"
url <- "https://api.openweathermap.org/data/2.5/weather"

query_params <- list(
  q = "Manaus",            # Capital do Amazonas
  appid = api_key,
  units = "metric",
  lang = "pt_br"
)

# ============================================================
# 2. REQUISIÇÃO À API E EXIBIÇÃO DOS DADOS CLIMÁTICOS
# ============================================================
resposta <- GET(url, query = query_params)

if (status_code(resposta) == 200) {
  dados <- content(resposta, as = "parsed")

  cidade    <- dados$name
  pais      <- dados$sys$country
  temp      <- dados$main$temp
  sensacao  <- dados$main$feels_like
  umidade   <- dados$main$humidity
  pressao   <- dados$main$pressure
  descricao <- dados$weather[[1]]$description
  vento     <- dados$wind$speed
  chuva     <- ifelse(is.null(dados$rain$`1h`), 0, dados$rain$`1h`)

  data_hora <- format(Sys.time(), "%d/%m/%Y %H:%M:%S")

  cat("\n============================================\n")
  cat("🌍 DADOS CLIMÁTICOS PARA", cidade, "-", pais, "\n")
  cat("🕒 Data/Hora:", data_hora, "\n")
  cat("🌡️ Temperatura:", temp, "°C\n")
  cat("🤔 Sensação térmica:", sensacao, "°C\n")
  cat("💧 Umidade:", umidade, "%\n")
  cat("📊 Pressão:", pressao, "hPa\n")
  cat("☁️ Condição:", descricao, "\n")
  cat("💨 Vento:", vento, "m/s\n")
  cat("🌧️ Chuva (última 1h):", chuva, "mm\n")
  cat("============================================\n\n")
} else {
  cat("❌ Erro na requisição. Código:", status_code(resposta), "\n")
}

# ============================================================
# 3. FUNÇÃO PARA ESTATÍSTICA BÁSICA (MÉDIA E DESVIO)
# ============================================================
calcular_estatisticas <- function() {
  cat("\n📊 ANÁLISE ESTATÍSTICA DOS DADOS DAS CULTURAS\n")
  cat("Instruções: Copie os valores numéricos do terminal do Python (ex: áreas ou quantidades)\n")
  cat("e cole abaixo, separados por espaço. Pressione Enter para finalizar.\n")

  # FORÇA A EXIBIÇÃO IMEDIATA DAS INSTRUÇÕES
  flush.console()

  # A MENSAGEM DE SOLICITAÇÃO AGORA FICA DENTRO DO readline()
  linha <- readline(prompt = "👉 Digite os números (separados por espaço): ")

  # Converte a string em um vetor numérico
  numeros <- as.numeric(unlist(strsplit(linha, " ")))

  # Remove valores NA (caso haja entradas inválidas)
  numeros <- numeros[!is.na(numeros)]

  if (length(numeros) == 0) {
    cat("❌ Nenhum número válido foi inserido.\n")
    return()
  }

  media <- mean(numeros)
  desvio <- sd(numeros)

  cat("\n--- RESULTADOS ---\n")
  cat("📌 Número de elementos:", length(numeros), "\n")
  cat("📊 Média:", round(media, 2), "\n")
  cat("📉 Desvio padrão:", round(desvio, 2), "\n")
}

# Chama a função
calcular_estatisticas()
```
